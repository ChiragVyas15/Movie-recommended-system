```python
#import libraries for data manipulation and analysis
import pandas as pd
import numpy as np
```

```python
#read the datasets
movies = pd.read_csv("tmdb_5000_movies.csv")
credits = pd.read_csv("tmdb_5000_credits.csv")
```

```python
# Merge the movies and credits datasets on title
movies = movies.merge(credits, on='title')
```

```python
movies.head(1)
```

```python
movies.info()
```

```python
#genre
#id
#keywords
#title
#overview
#cast
#crew
#we need these features to predict the movie
```

```python
#select only these features which are useful for the recommendation
movies=movies[[ 'id','genres', 'keywords', 'title', 'overview', 'cast', 'crew']]
```

```python
movies.head(1)
```

```python
#find any null value present in dataset or not
movies.isnull().sum()
```

```python
#if null value present than drop those rows
movies.dropna(inplace=True)
```

```python
movies.isnull().sum()
```

```python
#convert the genres column to a list of strings
#eval function converts the string representation of a list to an actual list
movies['genres'] = movies['genres'].apply(lambda x: [i['name'] for i in eval(x)] if pd.notnull(x) else [])
```

```python
movies.head()
```

```python
# Convert the keywords column to a list of strings using eval
movies['keywords']=movies['keywords'].apply(lambda x: [i['name'] for i in eval(x)] if pd.notnull(x) else [])
```

```python
movies.head()
```

```python
import ast
```

```python
#this function converts the string representation of a list of dictionaries to a list of movie cast
def convrt(obj):
    l=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter<3:
            l.append(i['name'])
            counter+=1
    return l
```

```python
movies['cast'] = movies['cast'].apply(convrt)
```

```python
movies['cast'].iloc[0]
```

```python
def convrt1(obj):
    l=[]
    
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            l.append(i['name'])
            break
    return l
```

```python
movies['crew']=movies['crew'].apply(convrt1)
```

```python
movies.head(1)
```

```python
movies['overview'].head()
```

```python
movies.head(1)
```

```python
movies['genres']=movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])#remove spaces in genres
```

```python
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])#remove spaces in keywords
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])#remove spaces in cast
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])#remove spaces in crew
movies['overview'] = movies['overview'].apply(lambda x: x.split())#convert overview to list of words
```

```python
movies.head()
```

```python
#concatenate all text columns into a single column
movies['tag']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
movies.head()
```

```python
#the final dataframe with selected columns which will be used for recommendations
df=movies[['id','title','tag']]
```

```python
df.head()
```

```python
#convert the tag column to a string using join 
df['tag']=movies['tag'].apply(lambda x: " ".join(x))
```

```python
df.head()
```

```python
#convert tag column string into lowercase
df['tag']=df['tag'].apply(lambda x: x.lower())
```

```python
df.head()
```

```python
#CountVectorizer is a technique used to convert a collection of text documents into a matrix of token counts.
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english') #5000 most frequent words will be considered and english stop words will be removed
```

```python
cv.fit_transform(df['tag']).toarray() #convert the text data into a matrix of token counts
```

```python
cv
```

```python
cv.get_feature_names_out() #get the feature names (words) from the CountVectorizer
```

```python
cv.get_feature_names_out()
```

```python
import nltk #import nltk library for natural language processing
```

```python
from nltk.stem.porter import PorterStemmer #stemming is a technique used in natural language processing to reduce words to their base or root form
stemmer = PorterStemmer()
```

```python
stemmer.stem('loving')
```

```python
#this stemming function reduces words to their root form this is apply on tag column
def stem_word(word):
    y=[]

    for i in word.split():
        y.append(stemmer.stem(i))
    return " ".join(y)
```

```python
df['tag'] = df['tag'].apply(stem_word)
```

```python
df.head()
```

```python
cv.get_feature_names_out()
```

```python
vector=cv.fit_transform(df['tag']).toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity #cosine similarity is a measure of similarity between two non-zero vectors of an inner product space
cosine_similarity=cosine_similarity(vector)
```

```python
#enumerate the movies based on their cosine similarity scores
sorted(list(enumerate(cosine_similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
#function to recommend similar movies
def recommand(movie):
    movie_index=df[df['title']==movie].index[0]
    similar_movies=sorted(list(enumerate(cosine_similarity[movie_index])),reverse=True,key=lambda x:x[1])[1:6]
    for i in similar_movies:
        print(df.iloc[i[0]]['title'])
```

```python
recommand("Avatar")
```
